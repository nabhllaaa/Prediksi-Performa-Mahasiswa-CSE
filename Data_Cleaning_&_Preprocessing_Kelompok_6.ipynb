```python
# Tahap 1 - Memuat Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_path = 'model_ready_data CSE Student.csv'
df = pd.read_csv(file_path)

print("✅ Tahap 1: Dataset Berhasil Dimuat")
print("-" * 50)

df.head()
```

```python
df.isnull().sum()

df = df.dropna()
```

```python
numerical_cols = ['Age', 'SleepPerDayHours', 'NumberOfFriend']

Q1 = df[numerical_cols].quantile(0.25)
Q3 = df[numerical_cols].quantile(0.75)
IQR = Q3 - Q1

df = df[~((df[numerical_cols] < (Q1 - 1.5 * IQR)) |
          (df[numerical_cols] > (Q3 + 1.5 * IQR))).any(axis=1)]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```
